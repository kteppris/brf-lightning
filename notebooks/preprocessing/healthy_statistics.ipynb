```python
from pathlib import Path
import numpy as np
from brf_lightning.data.bearing.utils import load_vibration_data

# Define dataset root and select normal-condition files
root_dir = Path("data/bearing_thl_v1")
csv_files = sorted(f for f in root_dir.glob("n*.csv"))  # only normal condition

all_signals = []

for file in csv_files:
    df, fs = load_vibration_data(file)
    x = df['sensor_bearing'].values.astype(np.float32)
    all_signals.append(x)

# Concatenate all normal condition signals
concatenated = np.concatenate(all_signals)

# Compute normalization stats
mean = np.mean(concatenated)
std = np.std(concatenated)

print(f"Mean (normal): {mean:.6f}")
print(f"Std (normal): {std:.6f}")
```
